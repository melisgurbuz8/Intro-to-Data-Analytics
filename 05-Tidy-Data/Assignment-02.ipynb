## Question 3

**True or False?**

> The average height of athletes who won a medal in Speed Skating is greater than the average height of athletes who won a medal in Basketball.

Write code to determine if this statement is True or False.

````python
# Your code goes here
medal_winners = athlete_events[athlete_events["Medal"].notna()]

speed_skating = medal_winners[medal_winners["Sport"] == "Speed Skating"]
basketball = medal_winners[medal_winners["Sport"] == "Basketball"]

speed_skating_height = speed_skating["Height"].mean()
basketball_height = basketball["Height"].mean()

speed_skating_height > basketball_height
````

## Question 4

Identify athletes who have participated in multiple Olympic events (more than one unique event). For these athletes:

- Calculate the sum of the total number of events they participated in altogether.
- Identify the athletes who have won at least two medals (any type of medal).  How many such athletes are there?

````python
# Your code goes here
events_per_athlete = athlete_events.groupby("ID")["Event"].nunique()
multiple_events = events_per_athlete[events_per_athlete > 1]
print(f'Sum of number of athletes who participated in multiple events is {multiple_events.sum()}')

medals_per_athlete = athlete_events.groupby("ID")["Medal"].count()
medals_per_athlete = medals_per_athlete[medals_per_athlete >= 2]
print(f'Sum of number of athletes who have won at least two medals is {medals_per_athlete.sum()}')
````

## Question 5

Identify all athletes who have won at least one medal and have participated in two or more different sports. For these athletes:

- Find the total number of different sports each of them participated in. Which athlete has the highest number of sports? And how many?
- Calculate the average number of medals won by these athletes. Which athlete has the maximum number of medals? And how many?

````python
# Your code goes here

sports_per_athlete = athlete_events.groupby("ID")["Sport"].nunique()

multiple_sports = sports_per_athlete[sports_per_athlete >= 2]

medals_per_athlete = athlete_events.groupby("ID")["Medal"].count()

medal_winners = medals_per_athlete[medals_per_athlete >= 1]

qualified_athletes = multiple_sports[
    multiple_sports.index.isin(medal_winners.index)
]

print(f'Athlete ID with the highest number of sports is {qualified_athletes.idxmax()}')
print(f'Highest number of sports is {qualified_athletes.max()}')

qualified_medals = medals_per_athlete[
    medals_per_athlete.index.isin(qualified_athletes.index)
]

print(f'Average number of medals is {qualified_medals.mean()}')

print(f'Athlete ID with the maximum number of medals is {qualified_medals.idxmax()}')
print(f'Maximum number of medals is {qualified_medals.max()}')
````

## Question 7

As part of a sponsorship deal, a company wants to endorse athletes who have a consistent performance in the Olympics. The company defines consistency for athletes who have participated more than one year and those who have won at least one medal in every year.

How many such athletes exist?



````python
# Your code goes here
years_per_athlete = athlete_events.groupby("ID")["Year"].nunique()

multiple_years = years_per_athlete[years_per_athlete > 1]

medal_years = athlete_events[
    athlete_events["Medal"].notna()
].groupby("ID")["Year"].nunique()

consistent_athletes = multiple_years[
    multiple_years.index.isin(medal_years.index)
]

consistent_athletes = consistent_athletes[
    consistent_athletes == medal_years[consistent_athletes.index]
]

print(f'Number of consistent athletes is {consistent_athletes.count()}')
````

## Question 8

A sports analytics firm wants to identify athletes who have shown versatility by competing in both individual and team sports. The firm is particularly interested in those who have won at least one medal in both types of events.

Provide the total number of athletes who have won medals in both individual and team sports.

Note: The following is a list of all team sports to reference for this question:

```
['Basketball', 'Football', 'Tug-Of-War', 'Ice Hockey', 'Handball', 'Water Polo', 'Hockey', 'Rowing', 'Bobsleigh', 'Sailing', 'Baseball', 'Softball',
'Rugby Sevens', 'Volleyball', 'Beach Volleyball', 'Synchronized Swimming', 'Curling', 'Lacrosse', 'Polo', 'Cricket', 'Military Ski Patrol', 'Croquet']
```

````python
# Your code goes here
team_sports = ['Basketball', 'Football', 'Tug-Of-War', 'Ice Hockey', 'Handball',
               'Water Polo', 'Hockey', 'Rowing', 'Bobsleigh', 'Sailing',
               'Baseball', 'Softball', 'Rugby Sevens', 'Volleyball',
               'Beach Volleyball', 'Synchronized Swimming', 'Curling',
               'Lacrosse', 'Polo', 'Cricket', 'Military Ski Patrol', 'Croquet']

medal_winners = athlete_events[athlete_events["Medal"].notna()]

team_medalists = medal_winners[
    medal_winners["Sport"].isin(team_sports)
]

individual_medalists = medal_winners[
    ~medal_winners["Sport"].isin(team_sports)
]

both = team_medalists[
    team_medalists["ID"].isin(individual_medalists["ID"])
]

print(both["ID"].nunique())
````

## Question 9

A sports health association is analyzing athletes' BMI to assess fitness, defining "fit" as those in the Normal weight category.

1. Create a **BMI** column using **Height** and **Weight**:  
$BMI=\frac{(Weight_{kg})}{Height_m^2}$

Fill missing Height and Weight values with average within the same Sex group and round BMI to 1 decimal place.

2. Count the number of "fit" athletes in the BMI range of 20-28 (inclusive).

Note: some athletes might have different weights and heights over the years. Take an average over all their measurements.

````python
# Your code goes here
male = athlete_events[athlete_events["Sex"] == "M"].copy()
female = athlete_events[athlete_events["Sex"] == "F"].copy()

male["Height"] = male["Height"].fillna(male["Height"].mean())
male["Weight"] = male["Weight"].fillna(male["Weight"].mean())

female["Height"] = female["Height"].fillna(female["Height"].mean())
female["Weight"] = female["Weight"].fillna(female["Weight"].mean())

athletes = pd.concat([male, female])

athlete_measurements = athletes.groupby("ID")[["Height", "Weight"]].mean()

athlete_measurements["BMI"] = (
    athlete_measurements["Weight"] /
    (athlete_measurements["Height"] / 100) ** 2
).round(1)

fit_athletes = athlete_measurements[
    (athlete_measurements["BMI"] >= 20) &
    (athlete_measurements["BMI"] <= 28)
]

print(fit_athletes["BMI"].count())
````
